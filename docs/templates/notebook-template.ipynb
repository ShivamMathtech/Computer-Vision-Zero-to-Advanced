## 4. Visual Explanation

[Replace: a labeled diagram or plot that directly explains the mechanism.]

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
image = np.array([[0, 20, 127, 255]], dtype=np.uint8)
plt.imshow(image, cmap="gray", vmin=0, vmax=255)
plt.title("Example input intensities")
plt.show()
```

## 5. Basic Implementation

[Replace: a small Python or NumPy operation that learners can trace.]

```python
inverted = (255 - image.astype(np.int16)).astype(np.uint8)
assert inverted.tolist() == [[255, 235, 128, 0]]
```

## 6. OpenCV Implementation

[Replace: relevant library comparison and explanation of its internal contract.]

```python
library_result = cv2.bitwise_not(image)
np.testing.assert_array_equal(library_result, inverted)
```

## 7. From-Scratch Implementation

[Replace: readable reference algorithm where educationally useful.]

```python
reference = np.empty_like(image)
for y in range(image.shape[0]):
    for x in range(image.shape[1]):
        reference[y, x] = 255 - int(image[y, x])
np.testing.assert_array_equal(reference, inverted)
```

## 8. Experiment

[Replace: one question, varied parameters, controls and predictions.]

```python
double_inversion = (255 - inverted.astype(np.int16)).astype(np.uint8)
np.testing.assert_array_equal(double_inversion, image)
```

## 9. Visualization

[Replace: before/after, features, masks, curves or other relevant evidence.]

```python
fig, axes = plt.subplots(2, 1, figsize=(7, 3))
for ax, values, title in zip(axes, [image, inverted], ["Input", "Inverted"], strict=True):
    ax.imshow(values, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
plt.tight_layout()
plt.show()
```
